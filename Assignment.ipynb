```python
import pandas as pd
```

```python
data = {
    'Name': ['Amit','Neha','Ravi','Sita'],
    'Age':[21,22,None,20],
    'Marks':[85,30,78,88]
}
df=pd.DataFrame(data)
df
```

```python
df['Grade']=['B','A','C','D']
df
```

```python
df[['Name','Marks']]
```

```python
df.isnull().sum()
```

```python
df['Age'].fillna(df['Age'].mean(),inplace=True)
df
```

```python
df['Age']=df['Age'].fillna(df['Age'].mean())
df
```

```python
df.loc[2,'Name']='Sonu'
df
```

```python
new_rows = pd.DataFrame([
    {'Name': 'Kumar', 'Age': 20, 'Marks': 90, 'Grade': 'A'},
    {'Name': 'Rahul', 'Age': 26, 'Marks': 85, 'Grade': 'B'}
])
```

```python
df = pd.concat([df, new_rows], ignore_index=True)
df
```
